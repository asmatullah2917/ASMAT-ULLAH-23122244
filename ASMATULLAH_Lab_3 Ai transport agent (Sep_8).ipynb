```python
!pip install -q -U google-genai

from google import genai

client = genai.Client(api_key="")

MODEL = "gemini-3.5-flash"


def transport_agent(user_input):

    prompt = f"""
I am an intelligent AI Transport Agent for Pakistan.

My job is to help users choose the most suitable transportation option
based on their budget, fare, seat preference, comfort, and travel requirements.

Analyze the user's request and provide a well-structured response with
clear headings:

1. Journey Details
   - Identify departure city
   - Destination city
   - Preferred travel requirements

2. Available Transport Options
   Compare suitable options such as:
   - Bus
   - Train
   - Flight
   - Ride-sharing or other local transport where appropriate

3. Fare Comparison
   Provide approximate fares in PKR.
   Clearly mention which option is:
   - Cheapest
   - Best value
   - Most comfortable

4. Seat Recommendations
   Recommend suitable seat options based on the user's preference, such as:
   - Window seat
   - Aisle seat
   - Front seat
   - Extra-legroom seat
   - Economy or Executive class

5. Best Transport Recommendation
   Select the most suitable option and explain why it is recommended.

6. Money-Saving Tips
   Give useful tips for finding cheaper fares and booking tickets.

Important:
- Do not claim that a seat is actually available unless real-time booking
  data is provided.
- Clearly label fares as approximate estimates when necessary.
- Ask for missing details such as departure city, destination, travel date,
  budget, or seat preference.
- Make the response easy to understand for students.

User Request:
{user_input}

Transport Recommendation:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("🚌 AI Transport Agent Started!")
print("I can help you choose suitable transport, fare, and seat preferences.")
print("Type 'exit' to stop.\n")


while True:

    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye! 👋")
        break

    try:
        reply = transport_agent(user)
        print("\n🤖 AI Transport Agent:\n")
        print(reply)
        print("\n" + "=" * 50 + "\n")

    except Exception as e:
        print("Error:", e)
```
